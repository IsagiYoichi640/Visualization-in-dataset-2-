```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.arange(1,11,1)
y1=(2*x)+1
y2=(2*x*x)+2
plt.plot(x,y1,'b', linewidth=3, label='y=2x+1')
plt.plot(x,y2,'r', linewidth=3, label='y=2x^2+2')
plt.legend()
plt.show()
```
