```python
import matplotlib.pyplot as plt
```

```python
x=[0,5,15,20,25,30,35,40, 45, 50]
y1=[10,15,25,20,18,35,15,0,20,25]
y2=[20,45,60,32,67,24,0,62,34, 20]
```

```python
plt.plot(x,y1,linestyle='dashed', marker='D')
plt.plot(x,y2,linestyle='dotted', marker='o')
plt.title("Velocity-Time Graph")
plt.xlabel("Time")
plt.ylabel("Velocity")
plt.legend(["y1", "y2"])
plt.show()
```

# Coclusion
* y2 is the fastest object as it has the highest velocity speeds.
* Between 30 and the 40 seconds, the velocity of the two object was at its lowest.
